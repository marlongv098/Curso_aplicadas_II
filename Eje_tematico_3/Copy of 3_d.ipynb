# Extremos de funciones

## Funciones diferenciables.

Una función $f$ definida en un disco abierto en el plano $x, y$ es dos veces
continuamente diferenciable en el disco si existen sus derivadas parciales $f_x$ y $f_y$
y tienen derivadas parciales continuas. Las derivadas parciales de $f_x$ y $f_y$ se denotan
 como $f_{xx}, f_{xy}$ y $f_{yx}, f_{yy}$ y se llaman segundas derivadas parciales. Una función diferenciable dos veces continuamente se llama función $C^2$.



## Derivadas Cruzadas

Para una función $f$ continuamente diferenciable dos veces  en un disco abierto en el plano $x, y$, las segundas derivadas parciales mixtas son iguales:
$$fxy = fyx.$$

## Punto extremo

Sea $f :D\subset \mathbb{R}^n→\mathbb{R}$. Se dice que $f$ tiene un mínimo local $f (A)$ en un punto $A$ en $D$ si hay una bola abierta en $\mathbb{R}^n$ con centro en $A$ de modo que para cada punto $X$ en $D$ que está en la bola, $$f (X) ≥ f (A)$$ $f$ tiene un máximo local $f (A)$ en un punto $A$ en $D$ si hay una bola abierta en $\mathbb{R}^n$ con centro en $A$ de modo que para cada punto $X$ en $D$ que está en la bola,
$$f (X) ≤ f (A)$$
Un punto que es un *máximo local o un mínimo local * se llama punto local extremo de $f$ en $D$. Decimos que un máximo local o un mínimo local es estricto si la desigualdad estricta correspondiente $f (X) > f (A)$ o $f (X) < f (A)$ se cumple excepto para $X = A$.


## Calculo puntos extremos

Sea $f$ una función diferenciable  $f :D\subset \mathbb{R}^n→\mathbb{R}$. Supongamos $f$ tiene un extremo local $f (A)$ en el punto interior $A$. Entonces las primeras  derivadas parciales  de $f$ son cero en $A$, es decir, $$∇f(A) = 0.$$

## Plano tangente

El cálculo de extremos en varias variables usa el conceptos de plano tangente a la superficie en un punto. En particular el punto de tangencia es el extremo de la función. Es decir, Los planos tangentes a considerar son los horizontales.

# Matriz Hessiana

Dada una función $C^2$  $f :\mathbb{R}^n→\mathbb{R}$ la matriz $n \times n$ de segundas derivadas parciales en $(x_1, . . . , x_n)$

$$H(x)=\begin{bmatrix}
f_{x_1x_1} & f_{x_1x_2} & \dots & f_{x_1x_n}\\
f_{x_2x_1} & f_{x_2x_2} & \dots & f_{x_2x_n}\\
\vdots & \vdots & \ddots & \vdots\\
f_{x_nx_1} & f_{x_nx_2} & \dots & f_{x_nx_n}
\end{bmatrix}
$$

es llamada la matriz *Hessiana* de $f$ en $x$.



## Máximos o Mínimos

Supongamos que $f$ es una función $C^2$ en el interior de $D ⊂ \mathbb{R}^{2}$, y que en algún punto interior $(c,d)$ de $D$,
$$∇f (c, d) = 0.$$

Sea $H(c,d)$ la matriz hessiana de segundas derivadas parciales de $f$ en $(c,d)$,
$$H(c,d)=\begin{bmatrix}
f_{x_1x_1}(c,d) & f_{x_1x_2}(c,d)\\
f_{x_1x_1}(c,d) & f_{x_2x_2}(c,d)
\end{bmatrix}
$$

* Si $H(c,d)$ es definida positiva, entonces $f$
tiene un mínimo local estricto en $(c,d)$. Esto es, si $f_xx>0$ y $\det H>0$.
* Si $H (c,d)$ es definida negativa, entonces $f$ tiene un máximo local estricto en $(c,d)$. Esto es, si $f_xx<0$ y $\det H>0$.
* Si $H (c,d)$ es indefinida, entonces $f$ no tiene un máximo local ni un mínimo local en ($c, d)$. Esto es, si $\det H<0$. Estos puntos se denominan *puntos silla*.

## Multiplicador de Lagrange

Sean $f$ y $g$ funciones $C^1$ de un conjunto abierto en $\mathbb{R}^2$ a $\mathbb{R}$, y enotamos por $S$ un conjunto de nivel $g(x, y) = c$. Sea $P$ un punto de $S$ tal que

* $∇g(P) \not=  0$, y

* $f$ tiene un extremo local en $S$ en $P$. Entonces existe un número $λ$ tal que

$$∇f (P) = λ∇g(P).$$
