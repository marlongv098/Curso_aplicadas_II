##<Font color="Green">Ejercicio 3.

* Considere un grupo de 5 personas, $P_1$, $P_2$, $P_3$, $P_4$ y $P_5$, que hacen parte de un equipo de trabajo. El coordinador del equipo ha observado las siguientes interacciones entre estas personas:

    *  $P_1$ se lleva bien con $P_2$, $P_3$ y $P_4$.
    * $P_2$ se lleva bien $P_1$ y $P_5$.
    * $P_3$ se lleva bien con $P_1$, $P_2$ y $P_4$.
    * $P_4$ Se lleva bien con $P_2$, $P_3$ y $P_5$.
    * $P_5$ se leva bien con $P_4$.

* Trace un digráfica $G$ que describa esta situación y Escriba la matriz de
adyacencia que representa a $G$. Es está digráfica Fuertemente Conexa?

**Ayuda**: Una digráfica $G$ es fuertemente conexa si para cualquier dos vértices distintos $P_i$ y $P_j$ eciste una trayectoria de $P_i$ a $P_j$ y una trayectoria de $P_j$ y $P_i$. En caso contrario, $G$ no es fuertemente conexa.  Tengan en cuenta el siguiente teorema:

**Teorema**: Una digráfica con $n$ vértices es fuertemente conexa si y sólo si su matriz de adyacencia $A(G)$ tiene la propiedad de que la matriz:   

$$\left [ A(G)) \right ]+\left [ A(G)) \right ]^{2}+...+\left [ A(G)) \right ]^{n-1}=E$$   

no tiene netradas iguales a cero.
