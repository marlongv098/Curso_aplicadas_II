*Ejemplo (Matriz de representación de un grafo)*.  Una relación $\mathcal{R}$ se puede ver como un gráfico dirigido, con nodos (o vértices) etiquetados $1,...,n$, y una arista dirigida de $j$ a $i$ para cada $(i,j)$. Esto se dibuja típicamente con un *grafo*, con flechas que indican la dirección de la arista, como se muestra en la figura 6.1 del texto de Boyd, para una relación sobre 4 objetos:

$$\mathcal{R} = \{(1, 2), (1, 3), (2, 1), (2, 4), (3, 4), (4, 1)\} $$

En general, una relación $\mathcal{R}$ sobre el conjunto $ \{1, 2, ...,n\} $v se puede representar con una matriz $A$ de $n\times n$ cuyas componentes vienen dadas  por

$$A_{ij}=\begin{cases}
1, & \text{ si } (i,j)\in \mathcal{R} \\
0, & \text{ si } (i,j)\not\in \mathcal{R}\end{cases}$$

Esta matriz se llama *matriz de adyacencia* asociada con el grafo. Dibujemos la matriz de adyacencia de la relación $\mathcal{R}$.



*Ejemplo (Matriz de datos de un grafo)*. Una colección de $n-$vectores, por ejemplo,  $n-$vectores de atributos asociados con $N$ objetos o personas, se pueden describir como una matriz de $n\times N$ cuyas $N$ columnas son los vectores. También es común describir esta colección de vectores utilizando la transpuesta de esa matriz. En ese caso, damos los vectores como una matriz $X$ de $N\times n$. Su $i-$ésima fila es $x^T_i$, la transpuesta del $i-$ésimo vector. Su $j-$ésima columna da el valor de la $j-$ésima entrada (o atributo) en la colección de $N$ vectores. Cuándo
un autor se refiere a una matriz de datos o matriz de atributos, por lo general se puede determinar
del contexto (por ejemplo, sus dimensiones) si se refieren a la matriz de datos organizados por filas o columnas.
